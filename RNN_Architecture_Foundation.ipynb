#### IMDB movie review sentiment classification dataset

```python
from keras.datasets import imdb
from keras.layers import Embedding , Flatten
```

```python
(x_train, y_train), (x_test, y_test) = imdb.load_data()
```

```python
print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
```

```python
x_train[0]  # already tokenized data
```

```python
print(len(x_train[0]))
```

```python
print(len(x_train[1]))
```

```python
x_train = pad_sequences(x_train, padding='post') # hyperparameter tuning
x_test = pad_sequences(x_test, padding= 'post')
```

```python
len(x_train[0])
```

```python
x_train = pad_sequences(x_train, padding='post', maxlen=50) # hyperparameter tuning
x_test = pad_sequences(x_test, padding= 'post', maxlen=50)
# to reduce the complexity hyperparameter tuning
# If we use pre padding, first 50 zeros will come only so always try to use post padding
```

```python
len(x_train[0])
```

```python
# Build the model
model = Sequential()
model.add(SimpleRNN(32, input_shape=(50,1), return_sequences=False))
model.add(Dense(1, activation='sigmoid'))
model.summary()
```

```python
# Build the model
model = Sequential()
model.add(SimpleRNN(32, input_shape=(50,1)))
model.add(Dense(1, activation='sigmoid'))
model.summary()
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
model.fit(x_train, y_train, epochs=5, batch_size=32, validation_data=(x_test, y_test))
```
